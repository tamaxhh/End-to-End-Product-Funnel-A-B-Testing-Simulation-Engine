```python
# 1. Get unique non-null brands per category
valid_brands = df.dropna(subset=["brand"])
unique_brand_per_cat = valid_brands.groupby("category_id")["brand"].nunique()

# 2. Filter for category_ids that have EXACTLY ONE unique brand
single_brand_cats = unique_brand_per_cat[unique_brand_per_cat == 1].index

# 3. Create mapping {category_id: brand_name}
mapping = (
    valid_brands[valid_brands["category_id"].isin(single_brand_cats)]
    .drop_duplicates(subset=["category_id"])
    .set_index("category_id")["brand"]
)

# 4. Fill NaNs using the mapping
df["brand"] = df["brand"].fillna(df["category_id"].map(mapping))
```

**Deterministic backfilling:**

```python
# 1. Build a lookup mapping of product_id -> brand from non-null rows
known_product_brands = (
    df.dropna(subset=["brand"])
    .drop_duplicates(subset=["product_id"])
    .set_index("product_id")["brand"]
)

# 2. Fill missing brand values using the product lookup
df["brand"] = df["brand"].fillna(df["product_id"].map(known_product_brands))
```

```python
# Fill remaining NaNs (multi-brand categories & 100% NaN categories) with "Unknown"
df["brand"] = df["brand"].fillna("Unknown")
```
